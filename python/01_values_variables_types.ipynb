# 01. 값, 변수, 자료형

Python 코드는 작은 값들을 이름 붙이고, 필요할 때 꺼내 쓰는 방식으로 움직입니다.

먼저 세 가지에 익숙해지면 좋습니다.

1. 값
2. 변수
3. 자료형

자료형은 쉽게 말해 값의 종류입니다. 실험 데이터를 정리하거나 분석 자동화 예제를 볼 때 `sample_name`, `condition`, `value` 같은 이름이 무엇을 가리키는지 읽을 수 있으면 충분히 큰 도움이 됩니다.

## 1. 값은 Python이 다루는 기본 재료입니다

Python에서 값은 코드가 다루는 실제 내용입니다. 예를 들어 숫자, 글자, 참/거짓 같은 것들이 값입니다.

````python
37
````

````python
0.42
````

````python
"control"
````

````python
True
````

위 값들은 각각 조금씩 성격이 다릅니다.

| 값 예시 | 의미 |
|---|---|
| `37` | 정수 |
| `0.42` | 소수 |
| `"control"` | 문자열, 즉 글자 데이터 |
| `True` | 참/거짓 값 |

생명과학 데이터에서도 이런 값들은 자주 등장합니다. 온도는 숫자, 샘플 이름은 문자열, 어떤 조건을 통과했는지는 참/거짓으로 표현할 수 있습니다.

## 2. 변수는 값에 붙이는 이름입니다

값을 매번 직접 쓰면 불편합니다. 그래서 값에 이름을 붙여 둡니다. 이 이름을 변수라고 합니다.

````python
sample_name = "sample_1"
condition = "control"
temperature = 37
od600 = 0.42

print(sample_name)
print(condition)
print(temperature)
print(od600)
````

`=`는 “같다”라는 뜻이 아니라, 오른쪽 값을 왼쪽 이름에 저장한다는 뜻입니다.

```python
od600 = 0.42
```

이 코드는 `od600`이라는 이름에 `0.42`라는 값을 저장합니다.

## 3. 변수 이름은 의미 있게 짓는 것이 좋습니다

아래 두 코드는 모두 실행됩니다.

````python
x = 0.42
y = 24

print(x)
print(y)
````

````python
od600 = 0.42
time_hour = 24

print(od600)
print(time_hour)
````

둘 다 Python 입장에서는 문제가 없습니다. 하지만 사람이 읽기에는 두 번째 코드가 훨씬 낫습니다.

실제 작업에서는 코드를 처음부터 새로 만드는 일보다, 이미 있는 코드를 읽고 필요한 부분을 바꾸는 일이 더 많습니다. 그래서 변수 이름이 무엇을 가리키는지 읽는 습관이 중요합니다.

## 4. 변수 값은 바뀔 수 있습니다

같은 변수에 새 값을 저장하면 이전 값은 새 값으로 바뀝니다.

````python
condition = "control"
print(condition)

condition = "treatment"
print(condition)
````

이런 특징 때문에 변수는 실험 조건, 파일 이름, 선택한 샘플 번호처럼 중간에 바꿔가며 사용할 수 있습니다.

## 5. 자료형 확인하기

Python은 값의 종류를 구분합니다. 이 값의 종류를 자료형이라고 부릅니다. 예를 들어 계산할 수 있는 숫자와 이름을 담는 글자는 서로 다르게 다뤄집니다.

`type()`을 사용하면 Python이 그 값을 어떤 자료형으로 보고 있는지 확인할 수 있습니다.

````python
print(type(37))
print(type(0.42))
print(type("control"))
print(type(True))
````

자주 보는 자료형은 이 정도입니다.

| 자료형 | 의미 | 예시 |
|---|---|---|
| `int` | 정수 | `37` |
| `float` | 소수 | `0.42` |
| `str` | 문자열 | `"control"` |
| `bool` | 참/거짓 | `True`, `False` |

처음부터 `int`, `float`, `str`, `bool` 같은 이름을 모두 외울 필요는 없습니다. 다만 숫자와 문자열은 서로 다르게 취급된다는 점은 기억해두면 좋습니다.

## 6. 숫자는 계산할 수 있습니다

````python
initial_od = 0.05
final_od = 0.45

increase = final_od - initial_od
print(increase)
````

````python
fold_change = final_od / initial_od
print(fold_change)
````

숫자 변수는 일반 계산처럼 더하고, 빼고, 곱하고, 나눌 수 있습니다.

## 7. 문자열은 이름이나 설명을 담을 때 사용합니다

````python
sample_name = "sample_1"
condition = "control"

message = sample_name + " is in " + condition
print(message)
````

문자열끼리는 `+`로 이어 붙일 수 있습니다. 하지만 변수 값을 문장 안에 넣을 때는 아래처럼 쓰는 방식이 더 편합니다.

````python
sample_name = "sample_1"
condition = "control"
od600 = 0.42

message = f"{sample_name} / {condition} / OD600 = {od600}"
print(message)
````

문자열 앞에 `f`를 붙이고, 문장 안에서 바뀌어 들어갈 값을 `{}` 안에 적습니다. 이름을 외우기보다 “문장 안에 변수 값을 끼워 넣는 방법” 정도로 이해하면 충분합니다.

## 8. 참/거짓 값은 조건 판단에 사용됩니다

````python
od600 = 0.42
is_high_growth = od600 > 0.3

print(is_high_growth)
print(type(is_high_growth))
````

`od600 > 0.3`은 “OD600 값이 0.3보다 큰가?”를 묻는 코드입니다. 결과는 `True` 또는 `False`가 됩니다.

이 값은 나중에 조건문에서 사용합니다.

## 9. 직접 고쳐보기

아래 셀에서 값만 바꿔 실행해 보세요.

- `sample_name`을 다른 이름으로 바꿔 보기
- `condition`을 `"treatment"`로 바꿔 보기
- `od600` 값을 바꿔 보기

````python
sample_name = "sample_1"
condition = "control"
od600 = 0.42

message = f"{sample_name} was measured under {condition}. OD600 = {od600}"
print(message)
````
